# Model 1: Target scaling linear baseline

## Import packages

```python
# Data loading packages
import anndata as ad
import numpy as np
import pandas as pd
import scanpy as sc
import scipy
```

## Load data

```python
context_A_data = sc.read_h5ad('../data/controls/context_A.h5ad')
context_B_data = sc.read_h5ad('../data/controls/context_B.h5ad')
context_C_data = sc.read_h5ad('../data/controls/context_C.h5ad')
```

```python
gene_names = pd.read_csv('../data/controls/gene_names.csv')
pert_counts = pd.read_csv('../data/controls/pert_counts.csv')
```

```python
# Convert to list so can pass into model methods
gene_names_list = list(gene_names['gene_name'])
pert_genes_list = list(pert_counts['target_gene'])
```

## Import and load baseline model

```python
from model import Model
```

```python
model_obj = Model()
```

# Calculate predictions for contexts A, B, C

```python
context_A_predictions = model_obj.predict_profiles(
    context_A_data,
    gene_names_list,
    pert_genes_list
)
context_A_predictions
```

```python
context_B_predictions = model_obj.predict_profiles(
    context_B_data,
    gene_names_list,
    pert_genes_list
)
context_B_predictions
```

```python
context_C_predictions = model_obj.predict_profiles(
    context_C_data,
    gene_names_list,
    pert_genes_list
)
context_C_predictions
```

# Compile all predictions

```python
combined_predictions = ad.concat([
    context_A_predictions,
    context_B_predictions,
    context_C_predictions
], axis = 0, label = "context", keys = ["A", "B", "C"])
combined_predictions.obs_names_make_unique()
```

```python
combined_predictions.write_h5ad('baseline_predictions.h5ad')
```
